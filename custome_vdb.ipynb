```python
import os

from dotenv import load_dotenv
from openai import OpenAI
load_dotenv()
client = OpenAI(
    api_key=os.environ["EURI_API_KEY"],
    base_url="https://api.euron.one/api/v1/euri",
)

resp = client.embeddings.create(
    model="text-embedding-3-small",
    input=["The quick brown fox jumps over the lazy dog."],
)
print(f"Dimensions: {len(resp.data[0].embedding)}")
print(f"First 5 values: {resp.data[0].embedding[:5]}")
```

```python
a = client.embeddings.create(
    model="text-embedding-3-small",
    input=["My name is muzmmil"],
).data[0].embedding
a
```

```python
c = client.embeddings.create(
    model="text-embedding-3-small",
    input=["Najmin is civil engineer."],
).data[0].embedding
```

```python
import numpy as np
```

```python
a = np.array(a)
b = np.array(b)
c = np.array(c)
```

```python
inp = client.embeddings.create(
    model="text-embedding-3-small",
    input=["What muzmmil do"],
).data[0].embedding
inp = np.array(inp)
```

```python
dic_a = np.linalg.norm(inp - a)
dic_b = np.linalg.norm(inp - b)
dic_c = np.linalg.norm(inp - c)
```
